# Experiment 3: Average sale price of houses with more than four bedrooms
**Subject:** Fundamentals of Data Science – NumPy Lab

## Problem statement
**Scenario:** House data from a CSV file is imported into a NumPy array `house_data`. Each row is a house; columns are number of bedrooms, square footage and sale price.

**Question:** Using NumPy arrays and operations, find the average sale price of houses with more than four bedrooms.

## Aim
Filter an array with a **boolean mask** and compute the mean of the selected values.

## Concepts used
| Idea | Example |
|---|---|
| Column selection | `house_data[:, 0]` → all rows, column 0 (bedrooms) |
| Boolean mask | `mask = house_data[:, 0] > 4` → array of `True/False` |
| Filtering rows | `house_data[mask]` → only the rows where mask is `True` |
| Mean of a column | `house_data[mask, 2].mean()` |
| Loading CSV | `np.loadtxt("file.csv", delimiter=",", skiprows=1)` |

## Algorithm
1. Load the CSV into `house_data`.
2. Create a mask: bedrooms `> 4`.
3. Select the price column of the matching rows.
4. Calculate the mean (check that at least one house matched).

```python
import numpy as np
```

### Step 1 – Create a sample CSV and load it
We write a small CSV file first so this notebook runs anywhere. **In your lab, skip the writing part and just load your own file.**

```python
csv_text = '''bedrooms,sqft,price
3,1500,250000
5,2800,480000
4,2100,350000
6,3500,620000
2,900,150000
5,2600,455000
3,1700,275000
7,4200,750000
4,2000,340000
5,3000,500000
'''
with open("house_data.csv", "w") as f:
    f.write(csv_text)

house_data = np.loadtxt("house_data.csv", delimiter=",", skiprows=1)   # skiprows=1 skips the header line

print("Shape:", house_data.shape)
print("Columns: [bedrooms, sqft, price]")
print(house_data)
```

### Step 2 – Create the mask (bedrooms > 4)

```python
bedrooms = house_data[:, 0]
price    = house_data[:, 2]

mask = bedrooms > 4
print("Mask         :", mask)
print("Houses found :", mask.sum())
```

### Step 3 – Average price of those houses

```python
selected_prices = price[mask]
print("Prices of houses with > 4 bedrooms:", selected_prices)

if selected_prices.size > 0:
    avg_price = selected_prices.mean()
    print(f"Average sale price (> 4 bedrooms): {avg_price:,.2f}")
else:
    print("No houses with more than 4 bedrooms.")
```

### Complete solution (short version)

```python
import numpy as np

house_data = np.loadtxt("house_data.csv", delimiter=",", skiprows=1)
avg = house_data[house_data[:, 0] > 4, 2].mean()
print("Average price of houses with more than 4 bedrooms:", avg)
```

## Extra – compare with all houses

```python
print(f"Average price (all houses)       : {price.mean():,.2f}")
print(f"Average price (> 4 bedrooms)     : {price[mask].mean():,.2f}")
print(f"Average price (<= 4 bedrooms)    : {price[~mask].mean():,.2f}")    # ~ flips True/False
```

## Result / Conclusion
A boolean mask filters the rows with more than four bedrooms, and `mean()` on their price column gives the required average.
