# Experiment 1: Average score per subject (NumPy axis operations)
**Subject:** Fundamentals of Data Science – NumPy Lab

## Problem statement
**Scenario:** A class of students has marks stored in a NumPy array `student_scores`. Each row is a student; columns are Math, Science, English, History (in this order).

**Question:** Calculate the average score for each subject and find the subject with the highest average. Assume a 4x4 matrix of marks.

## Aim
Use NumPy to compute column-wise averages and find the best subject.

## Concepts used
| Function | Purpose |
|---|---|
| `np.array()` | create the 2-D array |
| `arr.mean(axis=0)` | average **down the rows** → one value per **column** (subject) |
| `arr.mean(axis=1)` | average **across the columns** → one value per **row** (student) |
| `np.argmax()` | index of the largest value |

> **axis rule:** `axis=0` collapses rows (gives column results), `axis=1` collapses columns (gives row results).

## Algorithm
1. Create the 4x4 `student_scores` array.
2. Calculate `mean(axis=0)` to get the average of each subject.
3. Use `np.argmax` to find the position of the highest average.
4. Map that position to the subject name and print.

```python
import numpy as np
```

### Step 1 – Create the data (rows = students, columns = Math, Science, English, History)

```python
student_scores = np.array([
    [78, 85, 72, 90],
    [88, 79, 91, 84],
    [65, 70, 68, 75],
    [92, 95, 89, 87],
])

subjects = np.array(["Math", "Science", "English", "History"])

print("Shape:", student_scores.shape)
print(student_scores)
```

### Step 2 – Average score of each subject (column-wise)

```python
subject_avg = student_scores.mean(axis=0)

for name, avg in zip(subjects, subject_avg):
    print(f"{name:<8}: {avg:.2f}")
```

### Step 3 – Subject with the highest average

```python
best_index = np.argmax(subject_avg)

print("Highest average subject :", subjects[best_index])
print("Its average score       :", subject_avg[best_index])
```

### Complete solution (short version)

```python
import numpy as np

student_scores = np.array([[78, 85, 72, 90],
                           [88, 79, 91, 84],
                           [65, 70, 68, 75],
                           [92, 95, 89, 87]])
subjects = ["Math", "Science", "English", "History"]

averages = np.mean(student_scores, axis=0)
print("Averages:", averages)
print("Best subject:", subjects[np.argmax(averages)])
```

## Extra – the same code works for the full class of 32 students
Only the array size changes (32 x 4). Here we generate random marks to prove it.

```python
rng = np.random.default_rng(0)
big_scores = rng.integers(40, 101, size=(32, 4))      # 32 students, 4 subjects, marks 40-100

big_avg = big_scores.mean(axis=0)
print("Shape          :", big_scores.shape)
print("Subject averages:", big_avg.round(2))
print("Best subject   :", subjects[np.argmax(big_avg)])
```

## Extra – average of each student (row-wise)

```python
student_avg = student_scores.mean(axis=1)
for i, a in enumerate(student_avg, start=1):
    print(f"Student {i}: {a:.2f}")
```

## Result / Conclusion
Column-wise averaging with `mean(axis=0)` gives the average of every subject, and `np.argmax` finds the subject with the highest average (here **History**, with an average of 84.0).
