# Experiment 2: Average price of all products sold
**Subject:** Fundamentals of Data Science – NumPy Lab

## Problem statement
**Scenario:** Sales data for the past month is stored in a NumPy array. Assume a 3x3 matrix where each row represents the sales of a different product.

**Question:** How would you find the average price of all the products sold in the past month?

## Aim
Find the overall average of all values in a 2-D array.

## Concepts used
| Function | Purpose |
|---|---|
| `np.mean(arr)` | average of **all** elements (no axis given) |
| `np.mean(arr, axis=1)` | average of each **row** (each product) |
| `arr.sum()`, `arr.size` | sum of elements and count of elements |

## Algorithm
1. Create the 3x3 sales array (each row = one product, each column = one sales record/price).
2. Call `np.mean()` without an axis → average of all 9 values.
3. (Optional) Use `axis=1` for the average of each product.

```python
import numpy as np
```

### Step 1 – Create the data

```python
# Each row = one product, each column = one sale's price
sales = np.array([
    [120.0, 135.0, 128.0],   # Product 1
    [ 80.0,  95.0,  88.0],   # Product 2
    [200.0, 210.0, 190.0],   # Product 3
])

print("Shape:", sales.shape)
print(sales)
```

### Step 2 – Average price of all products sold

```python
average_price = np.mean(sales)
print("Average price of all products sold:", round(average_price, 2))
```

### Step 3 – Verify manually (sum / count)

```python
print("Sum   :", sales.sum())
print("Count :", sales.size)
print("Sum / Count =", round(sales.sum() / sales.size, 2))
```

## Extra – average price of each product (row-wise)

```python
product_avg = sales.mean(axis=1)
for i, a in enumerate(product_avg, start=1):
    print(f"Product {i}: {a:.2f}")

print()
print("Most expensive product on average: Product", np.argmax(product_avg) + 1)
```

## Note on interpretation
The question says each row is a product's sales. Here every cell holds a **price**, so the overall average is `np.mean(sales)`.
If your matrix instead stored **quantity sold** with a separate price array, use a weighted average:

```python
prices   = np.array([10.0, 25.0, 40.0])
quantity = np.array([[5, 3, 2],
                     [1, 4, 6],
                     [2, 2, 3]])

weighted_avg = (prices * quantity.sum(axis=1)).sum() / quantity.sum()
print("Weighted average price:", round(weighted_avg, 2))
```

## Result / Conclusion
`np.mean()` on the whole 3x3 array returns the overall average price of every product sold.
