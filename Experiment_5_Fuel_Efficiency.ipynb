# Experiment 5: Average fuel efficiency and percentage improvement
**Subject:** Fundamentals of Data Science – NumPy Lab

## Problem statement
**Scenario:** Fuel efficiency (miles per gallon) of different car models is stored in a NumPy array `fuel_efficiency`.

**Question:** Use NumPy arrays and arithmetic operations to calculate the average fuel efficiency and the percentage improvement in fuel efficiency between two car models.

## Aim
Calculate the mean of an array and the percentage improvement between two elements.

## Formula
$$\text{Percentage improvement} = \frac{\text{New} - \text{Old}}{\text{Old}} \times 100$$

## Concepts used
| Function | Purpose |
|---|---|
| `np.mean(arr)` | average of all values |
| `arr[i]`, `arr[j]` | pick two car models by index |
| `np.max / np.min` | best and worst model |

## Algorithm
1. Create `fuel_efficiency` (one value per car model).
2. Calculate `np.mean`.
3. Choose two models: the *old* one and the *new* one.
4. Apply the improvement formula.

```python
import numpy as np
```

### Step 1 – Create the data (miles per gallon)

```python
fuel_efficiency = np.array([28.5, 32.0, 25.0, 35.5, 30.0])   # Model A, B, C, D, E
models = np.array(["Model A", "Model B", "Model C", "Model D", "Model E"])

for m, mpg in zip(models, fuel_efficiency):
    print(f"{m}: {mpg} mpg")
```

### Step 2 – Average fuel efficiency

```python
average_mpg = np.mean(fuel_efficiency)
print(f"Average fuel efficiency: {average_mpg:.2f} mpg")
```

### Step 3 – Percentage improvement between two models

```python
old_model = 0      # Model A (index 0)
new_model = 3      # Model D (index 3)

old = fuel_efficiency[old_model]
new = fuel_efficiency[new_model]

improvement = (new - old) / old * 100
print(f"{models[old_model]} = {old} mpg,  {models[new_model]} = {new} mpg")
print(f"Percentage improvement: {improvement:.2f}%")
```

### Complete solution (short version)

```python
import numpy as np

fuel_efficiency = np.array([28.5, 32.0, 25.0, 35.5, 30.0])

avg = fuel_efficiency.mean()
improvement = (fuel_efficiency[3] - fuel_efficiency[0]) / fuel_efficiency[0] * 100

print("Average mpg          :", round(avg, 2))
print("Improvement A -> D   :", round(improvement, 2), "%")
```

## Extra – reusable function and the best vs worst model

```python
def percent_improvement(old, new):
    return (new - old) / old * 100

worst = np.min(fuel_efficiency)
best  = np.max(fuel_efficiency)

print(f"Worst: {models[np.argmin(fuel_efficiency)]} ({worst} mpg)")
print(f"Best : {models[np.argmax(fuel_efficiency)]} ({best} mpg)")
print(f"Improvement worst -> best: {percent_improvement(worst, best):.2f}%")
```

## Extra – improvement of every model over the average (vectorised)
NumPy applies the formula to **all** elements at once – no loop needed.

```python
vs_average = (fuel_efficiency - average_mpg) / average_mpg * 100

for m, v in zip(models, vs_average):
    print(f"{m}: {v:+.2f}% compared to average")
```

## Result / Conclusion
`np.mean` gives the average fuel efficiency, and `(New − Old) / Old × 100` gives the percentage improvement between two models.
