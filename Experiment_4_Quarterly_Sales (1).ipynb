# Experiment 4: Total yearly sales and percentage increase
**Subject:** Fundamentals of Data Science – NumPy Lab

## Problem statement
**Scenario:** Quarterly sales for the past four quarters are stored in a 1-D NumPy array `sales_data`.

**Question:** Using NumPy arrays and arithmetic operations, calculate the total sales for the year and the percentage increase in sales from the first quarter to the fourth quarter.

## Aim
Use `np.sum()` and array indexing with the percentage-change formula.

## Formula
$$\text{Percentage increase} = \frac{Q4 - Q1}{Q1} \times 100$$

## Concepts used
| Function | Purpose |
|---|---|
| `np.sum(arr)` | total of all elements |
| `arr[0]`, `arr[-1]` | first and last element (Q1 and Q4) |
| `np.diff(arr)` | change between consecutive elements |

## Algorithm
1. Create `sales_data` with 4 values.
2. Total = `np.sum(sales_data)`.
3. Q1 = `sales_data[0]`, Q4 = `sales_data[3]`.
4. Apply the formula.

```python
import numpy as np
```

### Step 1 – Create the data

```python
sales_data = np.array([150000, 175000, 160000, 210000])   # Q1, Q2, Q3, Q4
print("Quarterly sales:", sales_data)
```

### Step 2 – Total sales for the year

```python
total_sales = np.sum(sales_data)
print("Total sales for the year:", total_sales)
```

### Step 3 – Percentage increase from Q1 to Q4

```python
q1 = sales_data[0]
q4 = sales_data[3]

percentage_increase = (q4 - q1) / q1 * 100
print(f"Q1 = {q1}, Q4 = {q4}")
print(f"Percentage increase from Q1 to Q4: {percentage_increase:.2f}%")
```

### Complete solution (short version)

```python
import numpy as np

sales_data = np.array([150000, 175000, 160000, 210000])

print("Total sales      :", sales_data.sum())
print("% increase Q1->Q4:", round((sales_data[-1] - sales_data[0]) / sales_data[0] * 100, 2), "%")
```

## Extra – quarter-over-quarter growth
`np.diff` gives the change between neighbouring quarters.

```python
changes = np.diff(sales_data)                    # Q2-Q1, Q3-Q2, Q4-Q3
growth_pct = changes / sales_data[:-1] * 100     # divide by the earlier quarter

for i, g in enumerate(growth_pct, start=1):
    print(f"Q{i} -> Q{i+1}: {g:+.2f}%")
```

## Result / Conclusion
Total yearly sales = sum of the four quarters, and the Q1→Q4 increase = `(Q4 − Q1) / Q1 × 100`.
